```python
import numpy as np
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']
epsilon = 0.2
N_play = 10
M = len(machines)

# Preset reward patterns for experiment
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}
indices = {'A': 0, 'B': 0, 'C': 0}

# Initialize N, R, and Q tables
N = np.zeros(M, dtype=int)
Q = np.zeros(M, dtype=float)
R_history = {m: [] for m in machines}

total_reward = 0

print("--- Part B: Initial Test Run (Play each machine once) ---")
for idx, machine in enumerate(machines):
    reward = reward_patterns[machine][indices[machine]]
    indices[machine] += 1

    N[idx] += 1
    R_history[machine].append(reward)
    total_reward += reward

    # Incremental Sample-Average Formula
    Q[idx] = Q[idx] + (reward - Q[idx]) / N[idx]

    print(f"Initial play with Machine {machine}: Reward = {reward}, Updated Q = {Q[idx]:.2f}")

print("\n--- Modified Epsilon-Greedy Strategy Plays ---")
for play in range(4, N_play + 1):
    r = random.uniform(0, 1)

    # Identify machine with highest Q-value
    id_max = np.argmax(Q)

    # Decision Step
    if r < epsilon:
        # Exploration: Select randomly except highest Q machine
        valid_indices = [i for i in range(M) if i != id_max]
        selected_idx = random.choice(valid_indices)
        action_type = "Exploration"
    else:
        # Exploitation: Select highest Q machine
        selected_idx = id_max
        action_type = "Exploitation"

    selected_machine = machines[selected_idx]

    # Obtain Reward
    if indices[selected_machine] < len(reward_patterns[selected_machine]):
        reward = reward_patterns[selected_machine][indices[selected_machine]]
        indices[selected_machine] += 1
    else:
        reward = int(input(f"Play {play} ({action_type}) with Machine {selected_machine}. Enter reward: "))

    # Update Participation and Q-values
    N[selected_idx] += 1
    R_history[selected_machine].append(reward)
    total_reward += reward

    Q[selected_idx] = Q[selected_idx] + (reward - Q[selected_idx]) / N[selected_idx]

    print(f"Play {play:2d} | [{action_type}] Selected: {selected_machine} | Reward: {reward} | Q-values: A={Q[0]:.2f}, B={Q[1]:.2f}, C={Q[2]:.2f}")

# Part F - Display Results
print("\n" + "="*40)
print("FINAL RESULTS SUMMARY")
print("="*40)
for i, m in enumerate(machines):
    print(f"Machine {m}: Played {N[i]} times | Rewards: {R_history[m]} | Final Q-value: {Q[i]:.2f}")

print(f"\nTotal Reward Collected: {total_reward}")
```
