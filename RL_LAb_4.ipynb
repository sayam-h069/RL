```python
import math
import numpy as np

# Part A - Machine Initialization
machines = ['A', 'B', 'C']
M = len(machines)
N_play = 10
c = 2.0  # Exploration constant

reward_patterns = {'A': [3, 4, 2, 5], 'B': [6, 7, 5, 8], 'C': [4, 3, 5, 4]}

reward_indices = {'A': 0, 'B': 0, 'C': 0}

N = np.zeros(M, dtype=int)
Q = np.zeros(M, dtype=float)
UCB = np.zeros(M, dtype=float)


def get_reward(machine_name):
    idx = reward_indices[machine_name]
    pattern = reward_patterns[machine_name]
    R_val = pattern[idx % len(pattern)]
    reward_indices[machine_name] += 1
    return R_val


# Part B - Create Knowledge Base (Initial Test Step)
print("--- Part B: Initial Baseline Testing ---")
t = 0
for i in range(M):
    m_name = machines[i]
    R_val = get_reward(m_name)
    N[i] = 1
    Q[i] = R_val
    t += 1
    print(f"Initial play on Machine {m_name}: Reward = {R_val}")

print(f"Initial N: {dict(zip(machines, N))}")
print(f"Initial Q: {dict(zip(machines, Q))}\n")

# Main 10-Play Loop using UCB
total_reward = 0

print("--- UCB Strategy Execution for 10 Plays ---")
for play in range(1, N_play + 1):
    t += 1  # Increment overall time step

    # Part C - UCB Calculation & Machine Selection
    for i in range(M):
        UCB[i] = Q[i] + c * math.sqrt(math.log(t) / N[i])

    selected_idx = np.argmax(UCB)
    selected_machine = machines[selected_idx]

    # Part D - Reward Collection
    R_val = get_reward(selected_machine)
    total_reward += R_val

    # Part E - Incremental Q-Value Update
    N[selected_idx] += 1
    Q[selected_idx] = Q[selected_idx] + (
        R_val - Q[selected_idx]
    ) / N[selected_idx]

    # Print Step Log
    print(
        f"Play {play:2d} | Selected: {selected_machine} | Reward: {R_val} | "
        f"UCBs: {dict(zip(machines, np.round(UCB, 2)))} | Updated Q({selected_machine}): {Q[selected_idx]:.2f}"
    )

# Part F - Display Results
print("\n--- Part F: Final Summary Tables ---")
print("Machine Participation Count (N):", dict(zip(machines, N)))
print("Final Q-Values (Q):", dict(zip(machines, np.round(Q, 2))))
print(f"Total reward accumulated during 10 user plays: {total_reward}")
```
