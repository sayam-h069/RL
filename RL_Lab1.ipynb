Question 1

```python
# Number of machines
n = int(input("Enter number of machines: "))

machines = []
Q = []
N = []

# Input machine names and rewards
for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))

    machines.append(name)
    Q.append(0)
    N.append(0)

    # Incremental update
    for R in rewards:
        N[i] += 1
        Q[i] = Q[i] + (R - Q[i]) / N[i]

# Display final Q values
print("\nEstimated Q values:")

for i in range(n):
    print("Machine", machines[i])
    print("Q =", round(Q[i], 2))

# Greedy selection
best = 0

for i in range(1, n):
    if Q[i] > Q[best]:
        best = i

print("\nGreedy Selected Machine:", machines[best])
```

Question 2

```python
n = int(input("Enter number of machines: "))

machines = []
Q = []
N = []

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))

    machines.append(name)
    Q.append(0)
    N.append(0)

    for R in rewards:
        N[i] += 1
        Q[i] = Q[i] + (R - Q[i]) / N[i]

print("\nFinal Q values:")

for i in range(n):
    print(machines[i], "=", round(Q[i], 2))

# Greedy selection
max_index = 0

for i in range(1, n):
    if Q[i] > Q[max_index]:
        max_index = i

print("Greedy Selected Machine:", machines[max_index])
```
