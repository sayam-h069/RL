Question 1

```python
machines_data = {
    'A': [3, 2, 4],
    'B': [6, 7],
    'C': [5, 4, 6]
}
def arithmetic_mean_greedy(data):
    q_values = {}
    for machine, rewards in data.items():
        q_values[machine] = sum(rewards) / len(rewards)
    best_machine = max(q_values, key=q_values.get)
    return q_values, best_machine

def incremental_step_greedy(data):
    q_values = {}
    for machine, rewards in data.items():
        Q = 0.0
        for n, R in enumerate(rewards, start=1):
            Q = Q + (R - Q) / n
        q_values[machine] = Q
    best_machine = max(q_values, key=q_values.get)
    return q_values, best_machine

q_arithmetic, best_a = arithmetic_mean_greedy(machines_data)
q_incremental, best_i = incremental_step_greedy(machines_data)

print("--- Arithmetic Mean Method ---")
for m, q in q_arithmetic.items():
    print(f"Machine {m}: Estimated Reward = {q:.2f}")
print(f"Best Machine to Play: {best_a}\n")

print("--- Incremental Step Method ---")
for m, q in q_incremental.items():
    print(f"Machine {m}: Estimated Reward = {q:.2f}")
print(f"Best Machine to Play: {best_i}")
```

Question 2

```python
num_machines = int(input("Enter number of machines in cash: "))
machines = []

for i in range(num_machines):
    name = input(f"Enter name of machine {i + 1}: ")
    machines.append(name)

q_values = []

for machine in machines:
    n_plays = int(input(f"\nEnter number of plays for Machine {machine}: "))
    Q = 0.0

    for play in range(n_plays):
        reward = float(input(f"Enter reward for play {play + 1} on Machine {machine}: "))
        n = play + 1
        Q = Q + (reward - Q) / n

    q_values.append(Q)
print("\n--- Summary ---")
for i, machine in enumerate(machines):
    print(f"Machine {machine} Estimated Return: {q_values[i]:.2f}")

best_idx = q_values.index(max(q_values))
print(f"\nBest Machine to Play: {machines[best_idx]}")
```
