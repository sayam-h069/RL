```python
import numpy as np
import pandas as pd

states = ['S0', 'S1', 'Wall', 'G']
non_terminal_states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 1e-4

transition_table = {
    'S0': {'Up': ('S0', -1), 'Down': ('S1', -1), 'Left': ('S0', -1), 'Right': ('Wall', -1)},
    'S1': {'Up': ('S0', -1), 'Down': ('S1', -1), 'Left': ('S1', -1), 'Right': ('G', 10.0)},
}

print("=== Task 1 & 2: MDP Inspection & Transition Verification ===")
print(f"States: {states}")
print(f"Actions: {actions}")
print(f"Discount Factor (gamma): {gamma}")
print("\nTransitions:")
for s in non_terminal_states:
    for a in actions:
        next_s, r = transition_table[s][a]
        print(f"State: {s:2s} | Action: {a:5s} -> Next State: {next_s:4s} | Reward: {r}")

V = {s: 0.0 for s in states}
print("\n=== Task 3: Initialize State Values ===")
print(f"Initial V: {V}")

print("\n=== Task 4 & 5: Value Iteration ===")
iteration = 0
while True:
    iteration += 1
    V_new = V.copy()
    delta = 0.0

    for s in non_terminal_states:
        q_values = []
        for a in actions:
            next_s, r = transition_table[s][a]
            q_val = r + gamma * V[next_s]
            q_values.append(q_val)

        max_v = max(q_values)
        delta = max(delta, abs(max_v - V[s]))
        V_new[s] = max_v

    V = V_new.copy()
    print(f"Iteration {iteration:2d} | Delta: {delta:.6f} | V(S0): {V['S0']:.4f} | V(S1): {V['S1']:.4f}")

    if delta < theta:
        break

print("\n=== Task 6: Calculate Action Values (Q*) ===")
Q = {s: {} for s in non_terminal_states}
for s in non_terminal_states:
    for a in actions:
        next_s, r = transition_table[s][a]
        Q[s][a] = r + gamma * V[next_s]

q_df = pd.DataFrame(Q).T
print(q_df)

print("\n=== Task 7: Extract Optimal Policy ===")
policy = {}
for s in non_terminal_states:
    max_q = max(Q[s].values())
    best_actions = [a for a, q in Q[s].items() if abs(q - max_q) < 1e-6]
    policy[s] = best_actions
    print(f"State {s}: Optimal Action(s) -> {best_actions} with Q* = {max_q:.4f}")

print("\n=== Task 8: Verify Bellman Optimality ===")
for s in non_terminal_states:
    max_q = max(Q[s].values())
    v_val = V[s]
    print(f"State {s}: V*(s) = {v_val:.4f} | max_a Q*(s,a) = {max_q:.4f} | Match: {abs(v_val - max_q) < 1e-6}")
```
